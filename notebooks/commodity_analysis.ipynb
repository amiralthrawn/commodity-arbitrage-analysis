```python
brent_clean = brent.copy()
wti_clean = wti.copy()
gas_clean = gas.copy()
```

```python
brent_clean = brent_clean.dropna(subset=["brent_price"])
```

```python
wti_clean = wti_clean.dropna(subset=["wti_price"])
```

```python
gas_clean = gas_clean.dropna(subset=["natural_gas_price"])
```

```python
print("Brent :", brent_clean.shape)
print("WTI   :", wti_clean.shape)
print("Gas   :", gas_clean.shape)
```

```python
brent_clean = brent_clean[
    brent_clean["observation_date"] >= "2025-01-01"
]

wti_clean = wti_clean[
    wti_clean["observation_date"] >= "2025-01-01"
]

gas_clean = gas_clean[
    gas_clean["observation_date"] >= "2025-01-01"
]
```

```python
print("Brent :", brent_clean.shape)
print("WTI   :", wti_clean.shape)
print("Gas   :", gas_clean.shape)
```

```python
print("Brent :", brent_clean["observation_date"].min(), "→", brent_clean["observation_date"].max())
print("WTI   :", wti_clean["observation_date"].min(), "→", wti_clean["observation_date"].max())
print("Gas   :", gas_clean["observation_date"].min(), "→", gas_clean["observation_date"].max())
```

```python
brent_clean.to_csv(
    "../data/processed/brent_daily_clean.csv",
    index=False
)

wti_clean.to_csv(
    "../data/processed/wti_daily_clean.csv",
    index=False
)

gas_clean.to_csv(
    "../data/processed/natural_gas_daily_clean.csv",
    index=False
)
```
